```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make graphs look professional
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
```

```python
import pandas as pd

# Read the dataset directly using the file path
df = pd.read_csv("/content/london_weather.csv")

# Display the first five rows
df.head()
```

```python

# Dataset information
print("Dataset shape:", df.shape)

# Column information
print(df.info())

# Statistical summary
display(df.describe())

# Check missing values
print(df.isnull().sum())
```

```python
# create a Histogram
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="mean_temp",
    bins=30,
    kde=True
)

plt.title("Distribution of Average Daily Temperature in London")
plt.xlabel("Mean Temperature (°C)")
plt.ylabel("Number of Days")

plt.tight_layout()
plt.savefig("graph1_histogram.png", dpi=300)
plt.show()
```

```python
# Crate a Scatter plot
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="sunshine",
    y="max_temp",
    scatter_kws={"alpha": 0.3, "s": 15},
    line_kws={"color": "red"}
)

plt.title("Relationship Between Sunshine and Maximum Temperature")
plt.xlabel("Sunshine Hours")
plt.ylabel("Maximum Temperature (°C)")

plt.tight_layout()
plt.savefig("graph2_scatter.png", dpi=300)
plt.show()

# Calculate correlation
correlation = df["sunshine"].corr(df["max_temp"])

print("Correlation:", round(correlation, 3))
```

```python
# Create a correlation Heatmap

# Select numerical weather variables
weather_columns = [
    "sunshine",
    "max_temp",
    "mean_temp",
    "min_temp",
    "global_radiation",
    "precipitation",
    "pressure"
]

# Calculate correlations
correlation_matrix = df[weather_columns].corr()

# Create heatmap
plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of London's Weather Variables")

plt.tight_layout()
plt.savefig("graph3_heatmap.png", dpi=300)
plt.show()
```
